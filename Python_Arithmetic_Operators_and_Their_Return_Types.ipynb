**Python Arithmetic Operators and Their Return Types**

---



Operators themselves don't permanently "give float" or "give int".

The result depends on the operator and the types of the operands.

| Operator | Meaning             | Example   | Result | Data Type |
| -------- | ------------------- | --------- | ------ | --------- |
| `+`      | Addition            | `5 + 2`   | `7`    | `int`     |
| `+`      | Addition            | `5.0 + 2` | `7.0`  | `float`   |
| `-`      | Subtraction         | `5 - 2`   | `3`    | `int`     |
| `-`      | Subtraction         | `5.0 - 2` | `3.0`  | `float`   |
| `*`      | Multiplication      | `5 * 2`   | `10`   | `int`     |
| `*`      | Multiplication      | `5.0 * 2` | `10.0` | `float`   |
| `/`      | True Division       | `6 / 2`   | `3.0`  | **float** |
| `/`      | True Division       | `5 / 2`   | `2.5`  | **float** |
| `//`     | Floor Division      | `6 // 2`  | `3`    | `int`     |
| `//`     | Floor Division      | `5 // 2`  | `2`    | `int`     |
| `%`      | Modulus (Remainder) | `7 % 2`   | `1`    | `int`     |
| `**`     | Exponent (Power)    | `2 ** 3`  | `8`    | `int`     |


**The Special Case: /**

| Feature                                    | `/` (True Division)                            | `//` (Floor Division)                                                |
| ------------------------------------------ | ---------------------------------------------- | -------------------------------------------------------------------- |
| Purpose                                    | Divides and returns the exact result           | Divides and returns the floor (whole-number part)                    |
| Result Type (both operands are `int`)      | **Always `float`**                             | `int`                                                                |
| Result Type (if either operand is `float`) | `float`                                        | `float`                                                              |
| Keeps decimal part?                        | ✅ Yes                                          | ❌ No                                                                 |
| Suitable for `range()`?                    | ❌ No (returns `float`)                         | ✅ Yes (when both operands are `int`)                                 |
| Example                                    | `5 / 2`                                        | `5 // 2`                                                             |
| Output                                     | `2.5`                                          | `2`                                                                  |
| Example                                    | `6 / 2`                                        | `6 // 2`                                                             |
| Output                                     | `3.0`                                          | `3`                                                                  |
| Example                                    | `5.0 / 2`                                      | `5.0 // 2`                                                           |
| Output                                     | `2.5`                                          | `2.0`                                                                |
| Common Use Cases                           | Average, percentage, speed, exact calculations | Loop counts, indexing, page counts, splitting into equal whole parts |
